# AI Career Recommendation System - Exploratory Data Analysis

This notebook explores the structured career recommendation dataset, visualizing class distributions, feature correlations, and skill profiles.

```python
import pandas as pd
import os

# Load dataset
dataset_path = os.path.join('..', 'data', 'career_dataset.csv')
if os.path.exists(dataset_path):
    df = pd.read_csv(dataset_path)
    print(f'Successfully loaded dataset with {len(df)} rows.')
    print(df.head())
else:
    print('Dataset file not found at', dataset_path)
```

## Class Distribution

Let's see how many records we have for each career role.

```python
if 'df' in locals():
    print(df['career_role'].value_counts())
else:
    print('DataFrame not loaded')
```

## Experience Analysis

Let's check the distribution of years of experience.

```python
if 'df' in locals():
    print(df['years_of_experience'].describe())
else:
    print('DataFrame not loaded')
```
